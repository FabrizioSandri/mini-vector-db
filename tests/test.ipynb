```python
import random

import numpy as np
from mvdb import MVDB, Distance
```

```python
vdb = MVDB(dim=128, M=16, ef_construction=200, ef_search=50, max_level=10, distance=Distance.L2)
```

```python
rng = random.Random(42)

embeddings = [
    [rng.uniform(-1, 1) for _ in range(128)]
    for _ in range(10)
]
query = [rng.uniform(-1, 1) for _ in range(128)]
```

```python
for i, embedding in enumerate(embeddings):
    vdb.add(i, np.array(embedding, dtype=np.float32))
```

```python
vdb.size()
```

```python
results = vdb.search(np.array(query, dtype=np.float32), k=10)
results
```
